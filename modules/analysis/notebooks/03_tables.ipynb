## Table: paired tests across designs

D1 vs. D2 for each model, with the (workflow, smell) pair as the unit: McNemar's exact test for the two binary outcomes and a two-sided paired Wilcoxon signed-rank test for latency and cost, Holm-corrected across the four models of each outcome. The effect (Eff.) is D2 − D1: the difference of the two rates for the binary outcomes, and the median of the paired differences for latency (s) and estimated cost (USD × 1,000). All rep1, four models (Gemma 4 excluded), and the 570 pairs of the n15 ground truth. Each pair contributes one value per design:

- **Detection correctness** — 1 when the detection verdict matches the ground truth (the RQ1 scoring), over all 570 pairs.
- **End-to-end clean removal** — 1 when the run is Clean Removal, with the runs and rule of the end-to-end table: a detection miss counts as a failure, and `uncertain` runs are dropped. Only the positive pairs whose D1 and D2 runs are both assessable are compared.
- **Latency** — the run's `run_duration_seconds`.
- **Est. cost** — the run's estimated cost, with the runs and cost formula of the absolute overhead table.
